```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image=cv2.imread("D:\programming\pyhton\Morphological.png",0)

_, binary = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

kernal=np.ones((5,5),np.uint8)

opened=cv2.morphologyEx(binary,cv2.MORPH_OPEN,kernal)
closed=cv2.morphologyEx(binary,cv2.MORPH_CLOSE,kernal)


cv2.imshow("Original Image", image)
cv2.imshow("Transformed (Segmented) Image", closed)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.subplot(1,2,2)
plt.title("Transformed (Segmented) Image")
plt.imshow(closed, cmap='gray')

plt.show()



```
